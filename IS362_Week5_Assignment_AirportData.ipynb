This notebook explores three aviation-related questions using the `nycflights13` dataset:
**a.** What is the northernmost airport in the United States?  
**b.** What is the easternmost airport in the United States?  
**c.** On February 12th, 2013, which New York area airport had the windiest weather?
Sources are CSV files from [nycflights13/data-raw](https://github.com/hadley/nycflights13/tree/master/data-raw) and verify the results using external sources. 

Step 1: Import the Pandas library and load the CSV file to load airport and weather data

```python
>>> import pandas as pd
airports = pd.read_csv("https://raw.githubusercontent.com/hadley/nycflights13/master/data-raw/airports.csv")
weather = pd.read_csv("https://raw.githubusercontent.com/hadley/nycflights13/master/data-raw/weather.csv")
```

Step 2: Sort airports by their latitude descending order

```python
>>> northernmost_airports = airports.sort_values(by="lat", ascending=False).head(5)
northernmost_airports[["faa", "name", "lat", "lon"]]
```

# Answer a: Northernmost Airport
Based on the dataset, the northernmost airport is:
**Barrow Wiley Post–Will Rogers Memorial Airport (BRW)** in Alaska, latitude: 71.2854.
This matches external sources confirming BRW as the northernmost U.S. airport. 

Step 3: Sort airports by longitude ascending order

```python
>>> easternmost_airports = airports.sort_values(by="lon", ascending=True).head(5)
easternmost_airports[["faa", "name", "lat", "lon"]]
```

# Answer b: Easternmost Airport
Based on the dataset, the easternmost airport is: **Gerritsen Airport (BID)** on Block Island, Rhode Island, longitude: -71.5778. This aligns with external verification that Block Island is among the easternmost U.S. airport locations.

Step 4: Filter weather data for February 12, 2013 and group by origin and find max wind speed

```python
>>> weather["date"] = pd.to_datetime(weather[["year", "month", "day"]])
feb12_weather = weather[weather["date"] == "2013-02-12"]
windiest = feb12_weather.groupby("origin")["wind_speed"].max().sort_values(ascending=False)
windiest
```

# **Answer c: Windiest NYC Airport on February 12, 2013** 
The airport with the highest recorded wind speed on February 12, 2013 was **JFK (John F. Kennedy International Airport)**, and max wind speed was 24.6 mph. 
This was verified by checking for outliers and comparing across NYC airports (JFK, LGA, EWR).
